```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent=[
    "the glass of milk",
    "the glass of juice",
    "the cup of tea",
    "I am a good boy",
    "I am a good developer",
    "understand the meaning of words",
    "your videos are good"
]
```

```python
sent
```

```python
## Define the vocabulary size
voc_size=10000
```

```python
###One Hot Representation
one_hot_repr=[one_hot(words,voc_size) for words in sent]
one_hot_repr
```

You can see the vectors have been created using OHE and a vocabulary size of 10000. We will now pass the sentences one by one into the embedding layer for word-embedding representation

```python
## word embedding representation

from tensorflow.keras.layers import Embedding
# from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

**What is PAD_SEQUENCES:** The `pad_sequences` library helps us to ensure each sentences are represented in the same size. If not we cant train.

```python
sent_length=8
```

```python
embedded_docs=pad_sequences(one_hot_repr,padding='pre',maxlen=sent_length) ## This will pad the sentence with 0's at the start to complete a size of 8
print(embedded_docs)
```

```python
## feature representation
dim=10 ## just like in our example in the note, we used 300.
```

```python
model = Sequential()

model.add(Embedding(voc_size, 10, input_length=sent_length))

model.build(input_shape=(None, sent_length))

model.compile(optimizer='adam', loss='mse')
```

```python
model.summary()
```

What we just did is create the embedding layer which will be used to transform the words into vectors. NOTE: This is just a simple process we applied.

```python
## Now lets get them converted
model.predict(embedded_docs)
```

You can see every word is represented in 10 dimensions like we initialized.
